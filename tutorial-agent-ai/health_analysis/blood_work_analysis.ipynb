```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(model= "gemini-3.6-flash")
```

```python
with open('blood_work.txt', 'r') as f:
    blood_report = f.read()

print(blood_report[:500])
```

```python
extraction_prompt = f'''
You are a medical data extraction assistant.

From the blood report below, extract ALL test values and classify each one as HIGH, LOW and NORMAL
based on the reference ranges provided in the report.

Format your response as:
- Test Name : vlaue | Status : HIGH/LOW/NORML | Reference : range

Blood Report:
{blood_report}
'''
extraction_response = llm.invoke(extraction_prompt)
extracted_values = extraction_response.text

print("=== STAGE 1 : EXTRACTED VALUES ===")
print(extracted_values)
```

```python
diet_prompt = f'''
You are a clinical nutritionist specializing in Indian dietry habits.

Based on the blood work  analysis below, write:
1. A short health summary in 1 line explaning the paitient's condition in simple language.
2. A short, practical Indian diet plan having only two sections (1) Foods to avoid (2) Foods to eat more of.
    Do not include anyother in the diet plan.

Blood Work Analysis:
{extracted_values}
'''
diet_response = llm.invoke(diet_prompt)
diet_values = diet_response.text

print("=== STAGE 2: HEALTH SUMMARY & DIET PLAN ===")
print(diet_values)
```
