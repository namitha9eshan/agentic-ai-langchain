###### Loading API keys

```python
from dotenv import load_dotenv 
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_groq import ChatGroq


load_dotenv() #reads the .env file which contains the API keys
```

#### Calling Gemini Models

```python
llm = ChatGoogleGenerativeAI(model = "gemma-4-31b-it")

response = llm.invoke("What is Sri Lanka?")
print(response.text) #.text will filter out only the answer(metadata, reasoning is not included.)
```

```python
#system prompting
response = llm.invoke([
    ["system" , "You are an helpful AI that provide short one line answers"],
    ["human" , "what is Sri Lanka?"]])

print(response.text)
```

```python
#checking the available model list
from google import genai
client = genai.Client()
for m in client.models.list():
    print(m.name)
```

```python
#tempreture : Creativity of an AI 

llm = ChatGoogleGenerativeAI(model = "gemma-4-31b-it", temperature= 0)

prompt = "Give me a tagline for my coffee shop.Only the tagline, nothing else"

response = llm.invoke(prompt)
print(response.text)
```

```python
llm = ChatGoogleGenerativeAI(model = "gemma-4-31b-it", temperature= 2)

response = llm.invoke(prompt)
print(response.text)
```

#### Calling Groq Models

```python
from langchain_groq import ChatGroq
```

```python
from groq import Groq

client = Groq()

for m in client.models.list().data:
    print(m.id)
```

```python
llm = ChatGroq(model= "qwen/qwen3.8-27b", reasoning_format="parsed")

message = [('system', "Give me short answers only"),
           ("human", "How many moons does jupiter has?")]

response = llm.invoke(message)
print(response.content)
```
