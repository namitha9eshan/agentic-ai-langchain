###### Loading API keys

```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_groq import ChatGroq


load_dotenv() #reads the .env file 
```

```python
llm = ChatGoogleGenerativeAI(model = "gemma-4-31b-it")

response = llm.invoke("What is Sri Lanka?")
print(response.text) #.text will filter out only the answer(metadata, reasoning is not included.)
```

```python
#system prompting
response = llm.invoke([
    ["system" , "You are an helpful AI that provide short one line answers"],
    ["human" , "what is Sri Lanka?"]])

print(response.text)
```
